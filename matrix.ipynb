```python
import numpy as np

print("===================================")
print("      MATRIX OPERATIONS TOOL")
print("===================================")

# Input matrix size
rows = int(input("Enter number of rows: "))
cols = int(input("Enter number of columns: "))

print("\nEnter elements of Matrix A:")
A = np.array([[float(input(f"A[{i}][{j}] = ")) for j in range(cols)] for i in range(rows)])

print("\nEnter elements of Matrix B:")
B = np.array([[float(input(f"B[{i}][{j}] = ")) for j in range(cols)] for i in range(rows)])

while True:
    print("\n========== MENU ==========")
    print("1. Matrix Addition")
    print("2. Matrix Subtraction")
    print("3. Matrix Multiplication")
    print("4. Transpose of Matrix A")
    print("5. Determinant of Matrix A")
    print("6. Exit")

    choice = int(input("Enter your choice: "))

    if choice == 1:
        print("\nMatrix A:")
        print(A)
        print("\nMatrix B:")
        print(B)
        print("\nAddition Result:")
        print(A + B)

    elif choice == 2:
        print("\nSubtraction Result:")
        print(A - B)

    elif choice == 3:
        if cols == rows:
            print("\nMultiplication Result:")
            print(np.dot(A, B))
        else:
            print("\nMatrix multiplication requires compatible dimensions.")

    elif choice == 4:
        print("\nTranspose of Matrix A:")
        print(A.T)

    elif choice == 5:
        if rows == cols:
            print("\nDeterminant of Matrix A:")
            print(np.linalg.det(A))
        else:
            print("\nDeterminant can only be calculated for a square matrix.")

    elif choice == 6:
        print("\nThank you for using Matrix Operations Tool.")
        break

    else:
        print("\nInvalid Choice! Please try again.")
```
